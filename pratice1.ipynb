```python
inputString = "P@#yn26at^&i5ve"

letters = 0
digits = 0
symbols = 0

for char in inputString:
    if char.isalpha():
        letters += 1
    elif char.isdigit():
        digits += 1
    else:
        symbols += 1

print(f"Letters: {letters}, Digits: {digits}, Symbols: {symbols}")
```
